Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n, m = map(int, input().split())
matran = []
for i in range(n):
    hang = list(map(int, input().split()))
    matran.append(hang)
for hang in matran:
    for item in hang:
        print(f"{item:4d}", end="")
    print() 
```

W6A17
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
n = int(input())
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
duong_cheo_chinh = []
for i in range(n):
    duong_cheo_chinh.append(matrix[i][i])
duong_cheo_phu = [matrix[i][n - 1 - i] for i in range(n)]
print("Duong cheo chinh:", *duong_cheo_chinh)
print("Duong cheo phu:", *duong_cheo_phu)
```

W6A18
Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k) 

```python
n, m, k = map(int, input().split())
temp_list = []
for _ in range(n):
    row = tuple(map(int, input().split()))
    temp_list.append(row)
matrix_tuple = tuple(temp_list)
if 0 <= k < m:
    sum_col = sum(matrix_tuple[i][k] for i in range(n))
    print(sum_col)
if 0 <= k < n:
    sum_row = sum(matrix_tuple[k])
    print(sum_row)
```

W6A19
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def tim_vi_tri_lon_nhat(lst):
    if not lst:
        return None 
    vi_tri_max = 0 
    for i in range(1, len(lst)):
        if lst[i] > lst[vi_tri_max]:
            vi_tri_max = i 
    return vi_tri_max
lst = list(map(int, input().split()))
index = tim_vi_tri_lon_nhat(lst)
print(f"{index}")
```

W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
lst = list(map(int,input().split()))
k = int(input())
vtk=[]
for i in range(len(lst)):
    if lst[i] == k:
        vtk.append(i)
        break
if len(vtk) == 0:
    print(-1)
else:
    print(vtk)
    
```

W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
lst = input().split()
vi_tri_dict = {}
for idx, s in enumerate(lst):
    if s not in vi_tri_dict:
        vi_tri_dict[s] = [idx]
    else:
        vi_tri_dict[s].append(idx)
for s in vi_tri_dict:
    vi_tri_dict[s] = tuple(vi_tri_dict[s])
print(vi_tri_dict)
```

W6A22
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def sln():
    lst = list(map(int, input().split()))
    if not lst:
        return -1 
    max = 0 
    for i in range(1, len(lst)):
        if lst[i] > lst[max]:
            max = i 
    return max
print(sln())
```

W6A23 VIết hàm xóa đi các số <x có trong mảng

```python
def a():
    lst = list(map(int,input().split()))
    k = int(input())
    res = []
    for x in lst:
        if x >= k:
            res.append(x)
    return res
print(a())

```

W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
lst = list(map(int,input().split()))
tong = 0
for i in range(1,len(lst)-1):
    if (lst[i] > lst[i+1]) and (lst[i] > lst[i-1]):
        tong = tong + lst[i]
print(tong)
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không gi

```python
lst = list(map(int,input().split()))
tang = True 
giam = True 
for i in range(1,len(lst)):
    if lst[i] > lst[i - 1] :
        giam = False 
    if lst[i] < lst[i -1]  :
        tang = False 
if tang :
    print('Tang')
elif giam:
    print('Giam') 
else :
    print('Vo Danh')
```
