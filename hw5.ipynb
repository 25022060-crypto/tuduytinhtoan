Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số


```python
def is_armstrong(n):
    if n < 0: return False
    
    k = 0
    temp = n
    while temp > 0:
        k += 1
        temp //= 10
        
    total_sum = 0
    temp = n
    while temp > 0:
        digit = temp % 10
        total_sum += digit ** k
        temp //= 10
        
    return total_sum == n
n = int(input())
print(is_armstrong(n))
```

W5A18 VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên

```python
def a(n):
    temp = n
    reverse_n = 0
    while temp > 0:
        last_digit = temp % 10
        reverse_n = reverse_n * 10 + last_digit
        temp = temp // 10
    return n == reverse_n
n = int(input())
print(a(n))
```

W5A19 VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.

```python
def d(n):
    n = abs(n)
    while n > 9:
        sum_digits = 0
        while n > 0:
            sum_digits += n % 10  
            n //= 10              
        n = sum_digits            
    return n == 1
n = int(input())
print(d(n))
```

W5A27 Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def dem(Lst):
    count = 0
    for num in Lst:
        for digit in str(abs(num)):
            if digit != '0':
                count += 1
    return count
n = int(input())
Lst = [int(input()) for _ in range(n)]
print(dem(Lst))
```

W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).


```python

import math

def calculate_angle(x1, y1, x2, y2, x3, y3):
    ba = (x1 - x2, y1 - y2)
    bc = (x3 - x2, y3 - y2)
    dot_product = ba[0]*bc[0] + ba[1]*bc[1]
    mag_ba = math.sqrt(ba[0]**2 + ba[1]**2)
    mag_bc = math.sqrt(bc[0]**2 + bc[1]**2)
    res = math.degrees(math.acos(dot_product / (mag_ba * mag_bc)))
    return round(res, 2)
x1,y1,x2,y2,x3,y3 = map(int,input().split())
print(calculate_angle(x1, y1, x2, y2, x3, y3))
```
